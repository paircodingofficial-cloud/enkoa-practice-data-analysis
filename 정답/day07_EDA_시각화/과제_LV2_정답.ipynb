## 1. 지역권별 평균 요금 (병합 + 집계 + 정렬)
**배경**: 택시 승차 자치구(`pickup_borough`)만으로는 그 자치구가 도시의 어느 방향(중심부·동부·서부·북부)에 있는지 알기 어렵습니다. 자치구 설명표 `borough_info` 를 붙여 **지역권(`region`)** 별 평균 요금(`fare`)을 비교합니다.

**요구사항**:
- `data/taxis.csv` 파일을 읽어 변수 `df` 에, `data/borough_info.csv` 파일을 읽어 변수 `bi` 에 담으세요.
- 두 표를 `pickup_borough` 기준 **안쪽 병합**(`how='inner'`)한 뒤(6407행), `region` 별 **평균 `fare`** 를 구해 **내림차순으로 정렬**한 Series 를 `fare_by_region` 에 담으세요(인덱스는 지역권, 값은 평균 요금).
- `pickup_borough` 가 비어 있는 행(26건)은 자치구 정보가 없어 안쪽 병합에서 자연히 빠집니다.

**예시**
```
len(fare_by_region)               →  4
fare_by_region.index[0]           →  '동부'
round(fare_by_region.iloc[0], 2)  →  24.93
```
<details><summary>힌트</summary>

```text
접근방법:
- 자치구에 지역권 이름을 붙이려면 병합, 지역권별 평균은 그룹 집계다. 그 뒤 값 기준으로 정렬한다.

세부구현:
1. taxis 와 borough_info 를 pickup_borough 로 안쪽(inner) 병합한다
2. region 으로 묶어 fare 평균을 낸다
3. 평균값 기준 내림차순으로 정렬해 fare_by_region 에 담는다
```

</details>

````python
import pandas as pd

df = pd.read_csv('../../day07_EDA_시각화/data/taxis.csv')
bi = pd.read_csv('../../day07_EDA_시각화/data/borough_info.csv')
merged = df.merge(bi, on='pickup_borough', how='inner')
fare_by_region = merged.groupby('region')['fare'].mean().sort_values(ascending=False)
print(fare_by_region)
````

````python
# [자가채점]
assert len(fare_by_region) == 4
assert fare_by_region.index[0] == '동부', '평균 요금 1위 지역권은 동부(24.93)예요'
assert round(float(fare_by_region.iloc[0]), 2) == 24.93
assert list(fare_by_region.index) == ['동부', '북부', '서부', '중심부']
print("✅ 문제1 통과!")
````

### 해설 — 문제 1
- **접근법**: `merge(..., how='inner')` 로 자치구에 지역권·약칭을 붙인 뒤 `groupby('region')['fare'].mean()` 으로 지역권별 평균을 냅니다. 마지막에 `sort_values(ascending=False)` 로 큰 값부터 정렬해요.
- **흔한 실수**: `how='inner'` 라 자치구 결측 26행이 빠져 6407행이 됩니다(문제 의도와 일치). `how='left'` 로 두면 결측 자치구 행의 `region` 이 비어 집계 대상이 달라지니 주의하세요.
- **대안**: 정렬을 빼면 지역권 이름 순서로 남습니다. 값 순위를 보려면 `sort_values` 가 필요해요.

## 2. 결제수단·자치구 인원 교차표 (피벗 + 총계)
**배경**: 결제수단(`payment`)과 승차 자치구(`pickup_borough`)에 승차 건이 어떻게 나뉘어 있는지 **건수 교차표**로 봅니다.

**요구사항**:
- `data/taxis.csv` 파일을 읽어 변수 `df` 에 담으세요.
- `payment`(행) × `pickup_borough`(열) 로 **건수** 피벗을 만들되, `values='passengers'`, `aggfunc='count'`, `margins=True` 로 하여 `pv` 에 담으세요(개수를 세므로 `count`, `margins` 는 총계 행·열 `All` 을 붙입니다).
- 이어서 **같은 행·열 구성으로 `aggfunc` 만 바꿔** `fare` 의 **범위(최댓값 − 최솟값)** 격자표를 `pv_range` 에 담으세요(`margins` 없이, `lambda` 또는 `def` 로 작성. 결과는 2행 × 4열).
- 자가채점 아래 **서술 셀**에, 총계 셀 `pv.loc['All', 'All']` 이 무엇을 뜻하는지, 어느 결제수단·자치구 조합의 건수가 가장 많은지, 그리고 `pv_range` 에서 **현금 결제의 요금 범위가 자치구마다 크게 다른 것**이 무엇을 뜻할지 **3문장 이상** 적으세요.

**예시**
```
pv.loc['All', 'All']                       →  6364
pv.loc['credit card', 'Manhattan']         →  3839
pv_range.shape                             →  (2, 4)
round(pv_range.loc['cash', 'Queens'], 2)   →  149.0
round(pv_range.loc['cash', 'Bronx'], 2)    →  18.5
```
<details><summary>힌트</summary>

```text
접근방법:
- 행·열·값을 정하고 aggfunc 로 세는 것이 피벗테이블이다. margins 로 총계를 붙인다.

세부구현:
1. pivot_table 에 행 기준은 payment, 열 기준은 pickup_borough 를 준다
2. 값은 passengers, 세는 방식(aggfunc)은 개수(count)로 건수를 센다
3. 총계 옵션(margins)을 켜서 총계 행·열(All)을 붙여 pv 에 담는다
4. 같은 행·열로 다시 만들되 값은 fare, 집계는 '최댓값에서 최솟값을 뺀다'를 직접 써서 넘긴다
   (pivot_table 의 aggfunc 에도 groupby 처럼 내가 만든 함수를 넣을 수 있다 — 교안 4번 참고)
```

</details>

````python
import pandas as pd

df = pd.read_csv('../../day07_EDA_시각화/data/taxis.csv')
pv = df.pivot_table(index='payment', columns='pickup_borough',
                    values='passengers', aggfunc='count', margins=True)
print(pv)

# aggfunc 만 바꾸면 같은 격자에 다른 계산이 들어간다 (내가 만든 함수도 가능)
pv_range = df.pivot_table(index='payment', columns='pickup_borough',
                          values='fare', aggfunc=lambda s: s.max() - s.min())
print(pv_range.round(2))
````

````python
# [자가채점]
assert pv.loc['All', 'All'] == 6364, '총계 셀은 이 표가 집계한 전체 건수 6364 예요'
assert pv.loc['credit card', 'Manhattan'] == 3839
assert pv.loc['cash', 'Manhattan'] == 1397
assert pv_range.shape == (2, 4), 'margins 없이 2행 × 4열이 나와야 해요'
assert round(float(pv_range.loc['cash', 'Queens']), 2) == 149.0, '최댓값 - 최솟값이에요'
assert round(float(pv_range.loc['cash', 'Bronx']), 2) == 18.5
assert round(float(pv_range.loc['credit card', 'Bronx']), 2) == 79.36
print("✅ 문제2 통과!")
````

**관찰 (모범 서술 — 예시)**

총계 셀 `pv.loc['All', 'All']`(6364)은 결제수단과 자치구가 모두 기록된 승차 건을 전부 합한 값, 즉 이 교차표가 집계한 전체 건수를 뜻한다. 건수가 가장 많은 조합은 신용카드로 결제한 맨해튼 승차(3839건)로, 맨해튼에서 카드 결제가 압도적으로 많았음을 알 수 있다.

### 해설 — 문제 2
- **접근법**: `pivot_table` 의 `aggfunc='count'` 는 값(`passengers`)이 있는 행의 **개수**를 세어 건수 표를 만듭니다. `margins=True` 는 행·열 끝에 총계(`All`)를 덧붙여요.
- **흔한 실수**: 총계 셀 6364는 전체 6433건과 다릅니다 — `payment` 나 `pickup_borough` 가 비어 있는 행은 교차표에서 빠지기 때문입니다. 총계가 원본 행수와 다를 수 있음을 기억하세요.
- **대안**: 개수만 볼 거라면 `pd.crosstab(df['payment'], df['pickup_borough'], margins=True)` 로도 같은 표를 더 짧게 만들 수 있습니다.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../../day07_EDA_시각화/data/taxis.csv')
df['pickup'] = pd.to_datetime(df['pickup'])
df['hour'] = df['pickup'].dt.hour
dist_by_hour = df.groupby('hour')['distance'].mean()
plt.figure()
ax = sns.lineplot(data=dist_by_hour.reset_index(), x='hour', y='distance')
ax.set_title('시간대별 평균 이동거리')
plt.show()
````

### 해설 — 문제 4
- **접근법**: 문자열로 읽힌 날짜는 `pd.to_datetime` 으로 바꿔야 `.dt` 접근자로 시(`hour`)를 뽑을 수 있습니다. 시간대별 평균을 먼저 구해 선으로 그렸어요.
- **흔한 실수**: `pickup` 을 datetime 으로 바꾸지 않고 `.dt` 를 쓰면 에러가 납니다. 형 변환이 먼저예요.
- **대안**: `sns.lineplot(data=df, x='hour', y='distance')` 처럼 원본을 바로 넘겨도 되지만, 그러면 선 주위에 범위 띠가 그려집니다. 미리 평균을 구해 넘기면 깔끔한 선만 남아요.

## 5. 자치구별 요금 분포 비교 (바이올린 그림 + 관찰)
**배경**: 승차 자치구(`pickup_borough`)별로 요금(`fare`)이 얼마나 퍼져 있고, 비싼 요금 쪽으로 분포가 얼마나 길게 뻗는지 **바이올린 그림**으로 비교합니다.

````python
import pandas as pd

df = pd.read_csv('../../day07_EDA_시각화/data/taxis.csv')

def tip_ratio(s):
    """팁을 낸(tip > 0) 승객의 비율(%)."""
    return (s > 0).mean() * 100

fare_summary = df.groupby('pickup_borough').agg(
    평균요금=('fare', 'mean'),
    요금범위=('fare', lambda s: s.max() - s.min()),
    팁준비율=('tip', tip_ratio),
    주요결제수단=('payment', lambda s: s.mode()[0]),      # 범주형 -> 최빈값
    도착자치구_수=('dropoff_borough', 'nunique'),          # 범주형 -> 종류 수
).round(2)   # round 는 숫자 열만 건드리고 문자열 열은 그대로 둔다
print(fare_summary)
````

````python
# [자가채점]
assert fare_summary.shape == (4, 5), '자치구 4개 × 5열이 나와야 해요'
assert list(fare_summary.columns) == ['평균요금', '요금범위', '팁준비율', '주요결제수단', '도착자치구_수'], '열 이름을 지문 그대로 써 주세요'
assert sorted(fare_summary.index) == ['Bronx', 'Brooklyn', 'Manhattan', 'Queens'], '자치구 4개가 모두 있어야 해요 (순서는 상관없습니다)'
assert round(float(fare_summary.loc['Queens', '평균요금']), 2) == 24.93
assert round(float(fare_summary.loc['Manhattan', '요금범위']), 2) == 127.5, '범위는 최댓값 - 최솟값이에요'
assert round(float(fare_summary.loc['Manhattan', '팁준비율']), 2) == 69.7, '비율은 100을 곱한 퍼센트예요'
assert round(float(fare_summary.loc['Bronx', '팁준비율']), 2) == 5.05
assert fare_summary.loc['Manhattan', '주요결제수단'] == 'credit card', 'payment 의 최빈값이에요 (네 자치구 모두 credit card)'
assert int(fare_summary.loc['Manhattan', '도착자치구_수']) == 5, '맨해튼에서 출발한 승차의 도착 자치구는 5종류예요'
assert int(fare_summary.loc['Queens', '도착자치구_수']) == 4
print("✅ 문제10 통과!")
````

### 해설 — 문제 10
- **접근법**: 열마다 계산이 다르므로 **결과 열 이름을 직접 붙이는 집계**(`agg(새이름=('열', 함수))`)를 씁니다. 이름표(`'mean'`)와 `lambda`, `def` 로 만든 함수를 **한 번에 섞어** 넘길 수 있습니다.
- **흔한 실수**: 1) `agg(['mean', lambda s: ...])` 처럼 리스트로 넘기면 열 이름이 `<lambda_0>` 이 되어 자가채점이 찾는 `요금범위` 열이 없습니다. 2) `(s > 0).mean()` 에 **100을 곱하지 않아** 비율이 `0.697` 로 나오는 실수도 잦습니다.
- **범주형 두 열이 말해 주는 것**: `주요결제수단`은 **네 자치구 모두 `credit card`** 로 나옵니다 — 최빈값만 보면 네 곳이 똑같아 보이죠. 하지만 카드 비중은 퀸스 58.3%, 브롱스 74.8%로 꽤 다릅니다. **최빈값은 '무엇이 1등인가'만 말할 뿐 '얼마나 앞서는가'는 말해 주지 않습니다** — 교안 3절에서 본 그대로예요. `도착자치구_수`는 맨해튼만 5곳이고 나머지는 4곳입니다.
- **대안**: 비율도 `lambda s: (s > 0).mean() * 100` 한 줄로 쓸 수 있습니다. `def` 로 만들면 이름이 남아 다른 집계에서 다시 쓸 수 있죠.
- **읽을 거리**: 맨해튼은 평균 요금이 **가장 낮은데**(11.15) 팁을 낸 비율은 **가장 높습니다**(69.70%) — 평균 이동거리도 가장 짧고(2.35마일) 카드 결제가 73.3%라, 짧은 거리를 카드로 결제하는 승차가 많다는 뜻입니다.
- **주의 — 브롱스의 5.05%를 '현금 결제 탓'으로 읽으면 틀립니다**: 브롱스는 오히려 **카드 비중이 4개 자치구 중 1위(74.7%)** 이고, 카드 승차만 골라도 팁 기록률이 **6.8%** 에 그칩니다(맨해튼 95.6%). 게다가 브롱스 승차는 **99건 중 5건**뿐이라 비율이 크게 흔들립니다. 숫자 하나로 원인을 단정하지 말고 **표본 크기와 다른 설명 가능성**을 함께 보는 것이 교안_02 가 말한 좋은 EDA 입니다.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../../day07_EDA_시각화/data/taxis.csv')
tip_by_payment = df.groupby('payment')['tip'].mean()
count_by_borough = df.groupby('pickup_borough').size()
fare_by_payment = df.groupby('payment')['fare'].mean()
plt.figure()
ax = sns.barplot(data=fare_by_payment.reset_index(), x='payment', y='fare', errorbar=None)
ax.set_title('결제수단별 평균 요금')
plt.show()
print(tip_by_payment)
print(count_by_borough)
````
