## 2. 정규성·등분산 점검과 검정 선택 (가정 점검 + 서술)
**배경**: 세 종의 체중 평균을 비교하기 전에, 어떤 검정을 써야 할지 정하려면 **정규성**(Shapiro-Wilk)과 **등분산**(Levene)을 먼저 확인해야 합니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `body_mass_g`·`species` 열의 **결측 행을 제거**하세요(`dropna(subset=['body_mass_g', 'species'])`).
- 종별로 체중을 나눠 `adelie`·`chinstrap`·`gentoo` 에 담으세요(예: `adelie = df[df['species'] == 'Adelie']['body_mass_g']`).
- `pg.normality` 로 각 종 체중의 정규성을 검정해 **pval**(`['pval'].iloc[0]`)을 `shapiro_p_adelie`·`shapiro_p_chinstrap`·`shapiro_p_gentoo` 에 담으세요.
- `pg.homoscedasticity(data=df, dv='body_mass_g', group='species')` 로 등분산(Levene)을 검정해 **W**(`['W'].iloc[0]`, 검정통계량)를 `levene_stat`, **pval**(`['pval'].iloc[0]`)을 `levene_p` 에 담으세요.
- p값은 소수 **넷째 자리**, 검정통계량은 소수 **셋째 자리**까지 비교합니다.
- **종별 Q-Q Plot 도 함께** 그리세요 — `plt.subplots(1, 3, figsize=(15, 4.5))` 로 축 세 개를 만들고 각 종에 `pg.qqplot(..., dist='norm', ax=...)` 을 그린 뒤 제목을 답니다(자가채점 없음). **Shapiro 는 표본이 크면 사소한 편차에도 기각**하므로, 실제로 얼마나 벗어났는지는 그림으로 판단해야 합니다.
- 자가채점 아래 **서술 셀**에, 정규성·등분산 결과를 근거로 어떤 분산분석을 쓸지와 그 한계를 **2문장 이상** 적으세요.

**예시**
```
round(shapiro_p_adelie, 4)    → 0.0324
round(shapiro_p_chinstrap, 4) → 0.5605
round(shapiro_p_gentoo, 4)    → 0.2336
round(levene_stat, 3)         → 5.12
round(levene_p, 4)            → 0.0064
```
<details><summary>힌트</summary>

```text
접근방법:
- 종별로 체중을 나눈 뒤, 각 종에 pg.normality 로 정규성을, pg.homoscedasticity 로 등분산(Levene)을 확인한다.
- pg.normality 는 결과 표에서 pval 을, pg.homoscedasticity 는 W(검정통계량)·pval 을 꺼낸다.

세부구현:
1. penguins 를 불러와 체중·종의 결측 행을 제거한다
2. 종 이름으로 걸러 세 종의 체중 시리즈를 각각 만든다
3. 각 종에 pg.normality 를 돌려 ['pval'].iloc[0] 를 담는다
4. pg.homoscedasticity(data=df, dv='body_mass_g', group='species') 로 W·pval 을 담는다
5. 두 검정 결과를 근거로 어떤 분산분석이 적절한지 서술한다
```

</details>

````python
import pandas as pd
import pingouin as pg

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'species'])
# 세 종을 따로 떼어 낸다 — 정규성은 집단마다 확인해야 한다(합쳐서 보면 안 된다).
adelie = df[df['species'] == 'Adelie']['body_mass_g']
chinstrap = df[df['species'] == 'Chinstrap']['body_mass_g']
gentoo = df[df['species'] == 'Gentoo']['body_mass_g']
shapiro_p_adelie = pg.normality(adelie)['pval'].iloc[0]
shapiro_p_chinstrap = pg.normality(chinstrap)['pval'].iloc[0]
shapiro_p_gentoo = pg.normality(gentoo)['pval'].iloc[0]
# ANOVA 는 '세 집단의 분산이 비슷하다'를 전제로 한다 — 검정 전에 그 전제부터 확인한다.
hom = pg.homoscedasticity(data=df, dv='body_mass_g', group='species')
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, (name, vals) in zip(axes, [('Adelie', adelie), ('Chinstrap', chinstrap), ('Gentoo', gentoo)]):
    pg.qqplot(vals, dist='norm', ax=ax)
    ax.set_title(name + ' 체중 Q-Q')
plt.tight_layout(); plt.show()
print("세 종 모두 점들이 기준선을 꽤 잘 따른다 — Adelie 의 p=0.032 는 표본이 커서 나온 근소한 기각이다.")
levene_stat = hom['W'].iloc[0]
levene_p = hom['pval'].iloc[0]
print(round(shapiro_p_adelie, 4), round(shapiro_p_chinstrap, 4), round(shapiro_p_gentoo, 4))
print(round(levene_stat, 3), round(levene_p, 4))
````

````python
# [자가채점]
assert abs(shapiro_p_adelie - 0.0324) < 0.01
assert abs(shapiro_p_chinstrap - 0.5605) < 0.01
assert abs(shapiro_p_gentoo - 0.2336) < 0.01
assert abs(levene_stat - 5.12) < 0.01
assert abs(levene_p - 0.0064) < 0.01, 'levene(adelie, chinstrap, gentoo) 순서로 세 종을 함께 넣으세요'
print("✅ 문제2 통과!")
````

**검정 선택 (모범 서술 — 예시)**

정규성은 Chinstrap(p=0.5605)·Gentoo(p=0.2336)가 기각되지 않았고 Adelie(p=0.0324)만 근소하게 기각됐다. 다만 각 종의 표본이 크므로(68~151마리) 표본평균의 분포는 중심극한정리로 정규에 가까워져 분산분석은 이 정도 편차에 강건하다. 반면 Levene 등분산 검정은 p=0.0064 로 세 종의 분산이 서로 다르다고 본다. 등분산이 깨졌으므로 엄밀히는 Welch 분산분석(`pg.welch_anova`, 이 과정에서는 개념만 소개)이 더 안전하지만, 표본이 크고 세 집단 모두 충분히 크므로 이 실습에서는 일원분산분석(`pg.anova`)으로 진행하고 **등분산 위반**을 한계로 기록한다. 만약 정규성이 크게 깨졌다면 Kruskal-Wallis(`pg.kruskal`)를 고려했을 것이다.

### 해설 — 문제 2
- **접근법**: 세 집단 평균 비교의 가정은 1) 각 집단의 정규성(`pg.normality` — Shapiro-Wilk) 2) 집단 간 등분산(`pg.homoscedasticity` — Levene)입니다. 정규성은 집단마다 따로, 등분산은 긴 형태(`dv`·`group`) DataFrame 하나로 확인합니다.
- **흔한 실수**: `pg.normality` 결과 표에서 `pval` 열을, `pg.homoscedasticity` 에서 `W`(검정통계량)·`pval` 을 꺼냅니다. `pg.homoscedasticity` 는 긴 형태 데이터를 받으므로 세 종을 따로 나눠 넣지 말고 `data`·`dv`·`group` 으로 지정합니다.
- **대안·주의**: p>0.05 는 '정규성을 기각하지 못함'이지 '정규분포 확정'이 아닙니다. 표본이 크면 Shapiro 가 과민해지므로 Q-Q Plot 을 함께 보는 것이 좋습니다. 등분산이 깨지면 Welch 분산분석(`pg.welch_anova`)이 정석이지만 이 과정 범위 밖이라 개념만 소개합니다.

## 3. 종별 체중 차이 — 일원분산분석 + 효과크기 η² (검정)
**배경**: 앞에서 확인한 가정을 바탕으로, 세 종의 체중 평균이 서로 다른지 **일원분산분석(ANOVA)** 으로 검정하고 효과크기 **η²**(집단이 설명하는 분산 비율)까지 구합니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `body_mass_g`·`species` 결측 행을 제거해 `df` 에 담으세요(`dropna`).
- `pg.anova(data=df, dv='body_mass_g', between='species', detailed=True)` 로 일원분산분석 표를 만들어 `aov` 에 담고 `display(aov)` 하세요(긴 형태 — 값 열은 `dv`, 그룹 열은 `between`).
- 결과 표 첫 행에서 **F**(`aov['F'].iloc[0]`)를 `anova_f`, **p_unc**(`aov['p_unc'].iloc[0]`)을 `anova_p` 에 담으세요.
- **효과크기 η²** 는 결과 표의 **np2** 열(`aov['np2'].iloc[0]`)을 그대로 `eta_squared` 에 담으세요 — 일원분산분석에서 부분 η²(`np2`)가 곧 η² 입니다.
- 검정통계량·η² 는 소수 **셋째 자리**, p값은 소수 **넷째 자리**까지 비교합니다.

**예시**
```
round(anova_f, 3)     → 343.626
round(anova_p, 4)     → 0.0
round(eta_squared, 3) → 0.67
```
<details><summary>힌트</summary>

```text
접근방법:
- pg.anova 는 긴 형태(dv·between) 한 번으로 F·p·효과크기(np2)를 한 표로 준다.

세부구현:
1. penguins 를 불러와 체중·종 결측을 제거한다
2. pg.anova(data=df, dv='body_mass_g', between='species', detailed=True) 로 표 aov 를 만든다
3. aov['F']·['p_unc']·['np2'] 의 .iloc[0] 를 anova_f, anova_p, eta_squared 에 담는다
```

</details>

````python
import pandas as pd
import pingouin as pg

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'species'])
# detailed=True 라야 제곱합·자유도와 효과크기(np2)까지 표에 나온다.
#  p 값만으로는 '차이가 있다'까지고, 그 차이가 얼마나 큰지는 np2 가 말해 준다.
aov = pg.anova(data=df, dv='body_mass_g', between='species', detailed=True)
display(aov)
anova_f = aov['F'].iloc[0]
anova_p = aov['p_unc'].iloc[0]
eta_squared = aov['np2'].iloc[0]
print(round(anova_f, 3), round(anova_p, 4), round(eta_squared, 3))
````

````python
# [자가채점]
assert abs(anova_f - 343.626) < 0.01
assert abs(anova_p - 0.0) < 0.01
assert abs(eta_squared - 0.67) < 0.01
print("✅ 문제3 통과!")
````

### 해설 — 문제 3
- **접근법**: `pg.anova` 는 F=집단간분산/집단내분산 을 계산해 '적어도 한 집단의 평균이 다른가'를 검정하고, `np2`(부분 η²) 열로 효과크기까지 한 표에 줍니다. p≈0 이므로 세 종의 체중 평균이 모두 같다는 귀무가설을 기각합니다. η²=0.67 은 체중 분산의 약 67%를 종이 설명한다는 뜻으로 기준(0.01/0.06/0.14)으로 보면 매우 큰 효과입니다.
- **흔한 실수**: p값만 보고 끝내지 마세요. p는 '차이가 있나'만 답하고, **얼마나 큰 차이인지**는 효과크기(η²=`np2`)가 답합니다. `detailed=True` 를 빼면 `np2` 열이 안 나옵니다. 일원분산분석에서는 부분 η²(`np2`)가 곧 η²=집단간제곱합÷총제곱합 입니다.
- **대안**: `smf.ols('body_mass_g ~ C(species)', data=df).fit()` 뒤 `anova_lm` 으로도 같은 F·p 를 얻습니다. η² 는 소표본에서 과대추정되므로 논문에서는 ω²(오메가제곱)을 권하기도 합니다(개념).

## 4. 어느 종끼리 다른가 — Tukey 사후검정 (표)
**배경**: 분산분석은 '적어도 하나가 다르다'까지만 알려 줍니다. **어느 종 쌍**이 실제로 다른지 알려면 **Tukey HSD 사후검정**으로 모든 종 쌍을 비교해야 합니다.

> 문제 2 에서 **등분산이 깨졌다**는 것을 확인했죠(Levene p = 0.0064). 교안의 분기표대로면 정석은 **Games-Howell**(`pg.pairwise_gameshowell`)입니다. 실제로 돌려 보면 유의한 쌍이 **2개로 같아** 결론이 바뀌지 않습니다 — 그래서 이 실습은 Tukey 로 진행하되, 보고할 때 **등분산 위반**을 함께 적습니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `body_mass_g`·`species` 결측 행을 제거해 `df` 에 담으세요(`dropna`).

## 5. 가정이 깨졌다면 — Kruskal-Wallis (ANOVA 의 비모수 짝)
**배경**: 문제 2 에서 정규성·등분산을 점검하고, 문제 3 에서 **일원분산분석(ANOVA)** 으로 종별 체중 차이를 확인했습니다. 그런데 ANOVA 는 정규성·등분산을 전제합니다. 가정이 흔들릴 때 3집단 이상을 비교하는 비모수 대안이 **Kruskal-Wallis**(`pg.kruskal`)예요. 값 대신 **순위**를 쓰므로 가정에 덜 민감합니다. 같은 데이터에 돌려 보고 ANOVA 결론과 견줍니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `df` 에 담고, `body_mass_g` 의 **결측을 제거**하세요.
- `pg.kruskal(data=..., dv='body_mass_g', between='species')` 로 결과 표를 만들어 `display` 하세요.
- 그 표에서 검정통계량 `H` 를 `h_stat`, p-value `p_unc` 를 `kruskal_p` 에 담으세요.
- 종별 체중 **중앙값**을 `species_median`(Series, 종 이름이 인덱스)에 담으세요. Kruskal 은 평균이 아니라 순위를 보므로 중앙값이 짝이 되는 요약값입니다.
- **사후검정**: Kruskal 도 ANOVA 처럼 "어딘가 다르다" 까지만 말합니다. 어느 쌍이 다른지는 **Dunn 검정**으로 봅니다 — `sp.posthoc_dunn(df, val_col='body_mass_g', group_col='species', p_adjust='holm')` 로 쌍별 p 표를 만들어 `dunn` 에 담고 `display` 하세요(`import scikit_posthocs as sp`). 쌍이 여럿이라 **다중검정 보정**(`p_adjust='holm'`)이 필요합니다.
- 그 표에서 **유의한 쌍(p < 0.05)의 개수**를 `n_sig_dunn` 에 담으세요. 표가 대칭이고 대각선이 1.0 이므로 `int(((dunn < 0.05).sum().sum()) / 2)` 로 셉니다.
- `h_stat` 은 소수 둘째, `kruskal_p` 는 0 에 아주 가까운 값이라 `< 1e-40` 인지로 채점됩니다.

**예시**
```
round(h_stat, 2)          →  217.60
kruskal_p                 →  5.61e-48  (0 에 가까움)
species_median['Gentoo']  →  5000.0
n_sig_dunn                →  2        (문제 4 의 Tukey 와 같은 개수)
```
> **눈여겨볼 것**: 문제 3 의 ANOVA 도, 여기 Kruskal 도 **같은 결론**(종별 차이가 있다)에 이릅니다. 표본이 크고 차이가 뚜렷하면 두 방법이 잘 일치해요. 아래 서술에서 그 의미를 정리해 보세요.
<details><summary>힌트</summary>

```text
접근방법:
- 3집단 이상의 비모수 비교 함수에 데이터프레임·종속변수·집단변수를 이름으로 넘긴다.
- 순위 기반 검정이므로 요약값은 평균이 아니라 중앙값으로 본다.

세부구현:
1. read_csv 로 읽고 body_mass_g 결측을 제거한다(dropna 의 subset)
2. 비모수 3집단 검정 함수를 data·dv·between 인자로 호출해 결과 표를 받는다
3. 결과 표의 H 열과 p_unc 열 첫 값을 각각 h_stat·kruskal_p 에 담는다
4. species 로 묶어 body_mass_g 의 중앙값을 구해 species_median 에 담는다
5. 지문의 Dunn 함수로 쌍별 p 표를 만들어 dunn 에 담고, p<0.05 인 칸 수를 2로 나눠 n_sig_dunn 에 담는다
```

</details>

````python
import pandas as pd
import pingouin as pg
import scikit_posthocs as sp

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g'])
kw = pg.kruskal(data=df, dv='body_mass_g', between='species')
display(kw)
h_stat = kw['H'].iloc[0]
kruskal_p = kw['p_unc'].iloc[0]
species_median = df.groupby('species')['body_mass_g'].median()
print('H =', round(h_stat, 2), ', p =', format(kruskal_p, '.2e'))
print(species_median.to_string())

# 사후검정 — Dunn (Kruskal 의 짝). Holm 으로 다중검정 보정
dunn = sp.posthoc_dunn(df, val_col='body_mass_g', group_col='species', p_adjust='holm')
display(dunn.round(4))
n_sig_dunn = int(((dunn < 0.05).sum().sum()) / 2)
print('유의한 쌍 개수 =', n_sig_dunn, '(문제 4 의 Tukey 와 같다)')
````

````python
# [자가채점]
assert abs(h_stat - 217.60) < 0.01
assert kruskal_p < 1e-40
assert abs(species_median['Gentoo'] - 5000.0) < 0.01
assert abs(species_median['Adelie'] - 3700.0) < 0.01
assert n_sig_dunn == 2, 'Dunn 표에서 p<0.05 인 쌍의 개수예요 (Gentoo 가 다른 두 종과 다름)'
print("✅ 문제5 통과!")
````

**관찰 (모범 서술 — 예시)**

ANOVA 와 Kruskal-Wallis 가 모두 종별 체중 차이를 유의하다고 판정했다. 표본이 충분히 크고 Gentoo 와 나머지 두 종의 차이가 뚜렷해서, 평균을 쓰든 순위를 쓰든 같은 결론에 이른 것이다. 이렇게 두 방법이 일치하면 결론을 안심하고 쓸 수 있다. 반대로 표본이 작거나 이상치가 심해 **정규성이 크게 깨진 경우**, 또는 순서형 척도처럼 애초에 평균이 의미 없는 자료라면 Kruskal-Wallis 를 써야 한다.

### 해설 — 문제 5
- **접근법**: `pg.kruskal` 은 ANOVA 와 같은 **긴 형태(long-form)** 인자를 받습니다(`data`·`dv`·`between`). 반환 표의 열 이름은 ANOVA(`F`·`p-unc`)와 달라 **`H`·`p_unc`** 이니 주의하세요.
- **왜 중앙값인가**: Kruskal 은 값의 크기가 아니라 **순위**를 씁니다. 그래서 결과를 설명할 때는 평균보다 **중앙값**을 함께 제시하는 것이 자연스럽습니다(Gentoo 5000g vs Adelie·Chinstrap 3700g).
- **흔한 실수**: 세 종의 체중을 각각 배열로 뽑아 넘기려 하면 인자 형태가 맞지 않습니다. `pg.kruskal` 은 원본 데이터프레임 하나와 열 이름 두 개로 호출합니다.
- **한계**: ANOVA 와 마찬가지로 "적어도 한 쌍은 다르다" 까지만 말해 줍니다. 어느 쌍인지는 사후검정이 필요하고, 비모수라면 Tukey 가 아니라 **Dunn 검정**을 씁니다.
- **대안**: 정규성은 괜찮은데 **등분산만** 깨졌다면 Kruskal 대신 **Welch ANOVA**(`pg.welch_anova`)가 더 적절합니다.

## 6. 종과 서식지는 독립인가 — 카이제곱 + Cramér's V (검정)
**배경**: 펭귄의 **종(`species`)** 과 **서식지(`island`)** 가 서로 관련 있는지(독립이 아닌지) **카이제곱 독립성 검정**으로 확인하고, 관련의 세기를 효과크기 **Cramér's V** 로 잽니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `species`·`island` 결측 행을 제거하세요(`dropna(subset=['species', 'island'])` — 이 두 열은 원래 결측이 없어 행 수는 그대로 344 입니다).
- `pg.chi2_independence(data=df, x='species', y='island')` 로 카이제곱 독립성 검정을 실행하세요. 이 함수는 **(기대빈도표, 관측빈도표, 통계량표)** 세 개를 순서대로 돌려줍니다: `expected, observed, chi_stats = pg.chi2_independence(...)`. 관측 교차표는 `display(observed)` 로 확인하세요.
- 통계량표 `chi_stats` 에서 **표준 Pearson 검정 행**(`chi_stats[chi_stats['test'] == 'pearson']`)을 골라 `chi2`(→`chi2`), `pval`(→`chi_p`), `cramer`(→`cramers_v`) 를 꺼내세요. **Cramér's V 는 `cramer` 열로 자동 계산**됩니다(손계산 불필요).
- **사후분석**: 카이제곱은 "두 변수가 관련 있다" 까지만 말합니다. **어느 칸이 기대보다 두드러지는지**는 **조정된 잔차**로 봅니다 — `Table(observed.values).standardized_resids` 로 계산해 `observed` 와 같은 행·열 이름을 붙인 DataFrame `resid_df` 에 담고 `display` 하세요(`from statsmodels.stats.contingency_tables import Table`).
- **|조정된 잔차| > 2 인 칸의 개수**를 `n_strong` 에 담으세요(`int((resid_df.abs() > 2).sum().sum())`). 양수면 기대보다 **많다**, 음수면 **적다**는 뜻입니다.
- 검정통계량·Cramér's V 는 소수 **셋째 자리**, p값은 소수 **넷째 자리**까지 비교합니다.

**예시**
```
round(chi2, 3)      → 299.55
round(chi_p, 4)     → 0.0
round(cramers_v, 3) → 0.66
n_strong            → 8      # 9칸 중 8칸이 |잔차| > 2
```
> **읽는 법**: Gentoo × Biscoe 의 조정된 잔차가 **+14.1** 로 가장 큽니다 — Gentoo 는 Biscoe 섬에 **유독 몰려 있다**는 뜻입니다. Cramér's V = 0.66 이라는 강한 연관이 **구체적으로 어느 칸에서 오는지**를 잔차가 알려 줍니다.
<details><summary>힌트</summary>

```text
접근방법:
- 두 범주형 변수의 관련성은 pg.chi2_independence 로 확인한다 — (기대, 관측, 통계량) 세 표를 준다.
- 통계량표의 test=='pearson' 행이 표준 카이제곱이고, cramer 열이 효과크기다.

세부구현:
1. penguins 를 불러와 종·서식지 결측을 제거한다
2. expected, observed, chi_stats = pg.chi2_independence(data=df, x='species', y='island')
3. pearson = chi_stats[chi_stats['test'] == 'pearson'] 로 표준 행을 고른다
4. pearson['chi2']·['pval']·['cramer'] 의 .iloc[0] 를 chi2, chi_p, cramers_v 에 담는다
5. Table(관측표.values).standardized_resids 로 조정된 잔차를 구해 행·열 이름을 붙여 resid_df 에 담는다
6. 절댓값이 2를 넘는 칸 수를 세어 n_strong 에 담는다
```

</details>

````python
import pandas as pd
import pingouin as pg
from statsmodels.stats.contingency_tables import Table

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['species', 'island'])
expected, observed, chi_stats = pg.chi2_independence(data=df, x='species', y='island')
display(observed)                       # 관측 교차표
pearson = chi_stats[chi_stats['test'] == 'pearson']
chi2 = pearson['chi2'].iloc[0]
chi_p = pearson['pval'].iloc[0]
cramers_v = pearson['cramer'].iloc[0]
print(round(chi2, 3), round(chi_p, 4), round(cramers_v, 3))

# 사후분석 — 조정된 잔차: 어느 칸이 기대보다 두드러지나
resid_df = pd.DataFrame(Table(observed.values).standardized_resids,
                        index=observed.index, columns=observed.columns)
print('[조정된 잔차 — 양수=기대보다 많음]')
display(resid_df.round(2))
n_strong = int((resid_df.abs() > 2).sum().sum())
print('|잔차| > 2 인 칸 =', n_strong, '개 / 전체', resid_df.size, '칸')
````

````python
# [자가채점]
assert abs(chi2 - 299.55) < 0.01
assert abs(chi_p - 0.0) < 0.01
assert abs(cramers_v - 0.66) < 0.01, "통계량표 pearson 행의 cramer 열이 Cramér's V 예요"
assert n_strong == 8, '조정된 잔차의 절댓값이 2를 넘는 칸 수예요 (9칸 중 8칸)'
assert abs(resid_df.loc['Gentoo', 'Biscoe'] - 14.13) < 0.05, 'Gentoo x Biscoe 잔차가 가장 큽니다'
print("✅ 문제6 통과!")
````

### 해설 — 문제 6
- **접근법**: 카이제곱 독립성 검정은 교차표의 관측빈도와 '두 변수가 독립일 때 기대되는 빈도'의 차이를 봅니다. `pg.chi2_independence` 는 기대·관측·통계량 세 표를 주고, 통계량표의 `pearson` 행이 표준 카이제곱입니다. p≈0 이라 종과 서식지는 독립이 아니며, Cramér's V(`cramer`)=0.66 은 관련이 매우 강함(기준 0.1/0.3/0.5)을 뜻합니다. 관측 교차표를 보면 Chinstrap 은 Dream 에만, Gentoo 는 Biscoe 에만 살아 종이 서식지를 거의 결정합니다.
- **흔한 실수**: 통계량표에는 여러 검정(`pearson`·`log-likelihood` 등)이 함께 나옵니다 — 반드시 **`test=='pearson'` 행**을 골라야 scipy `chi2_contingency` 와 같은 값이 됩니다(3×3 표라 Yates 보정은 없습니다).
- **대안·보너스**: `cramer` 열이 Cramér's V 라 손계산이 필요 없습니다. 2×2 표에서 기대빈도가 5 미만인 칸이 많으면 카이제곱 대신 Fisher 정확검정(`scipy.stats.fisher_exact`)을 씁니다 — 여기서는 기대빈도가 충분해 카이제곱 가정이 충족됩니다.

## 7. 물갈퀴 길이로 체중 예측 — 단순 선형회귀 (검정 + 그래프)
**배경**: 물갈퀴 길이(`flipper_length_mm`)가 길수록 체중(`body_mass_g`)이 무거울까요? **단순 선형회귀**로 관계를 수식으로 요약하고, 회귀선을 산점도 위에 그려 봅니다.

### (1) 회귀 적합 — 계수·R²·p값 (자가채점 있음)
**요구사항**:
- `data/penguins.csv` 를 읽어 `body_mass_g`·`flipper_length_mm` 결측 행을 제거하세요(`dropna`).
- `smf.ols('body_mass_g ~ flipper_length_mm', data=df).fit()` 로 회귀모형을 적합해 `simple_model` 에 담으세요.
- 결정계수 `simple_r2 = simple_model.rsquared`, 물갈퀴 계수 `simple_coef = simple_model.params['flipper_length_mm']`, 그 계수의 p값 `simple_p = simple_model.pvalues['flipper_length_mm']` 를 각각 담으세요.
- 결정계수·계수는 소수 **셋째 자리**, p값은 소수 **넷째 자리**까지 비교합니다.

**예시**
```
round(simple_r2, 3)   → 0.759
round(simple_coef, 3) → 49.686
round(simple_p, 4)    → 0.0
```
<details><summary>힌트</summary>

```text
접근방법:
- smf.ols 에 '종속변수 ~ 독립변수' 식과 데이터를 넣고 fit 한다.
- 적합 결과 객체에서 rsquared, params, pvalues 로 값을 꺼낸다.

세부구현:
1. penguins 를 불러와 체중·물갈퀴 결측을 제거한다
2. ols 식 'body_mass_g ~ flipper_length_mm' 로 모형을 적합한다
3. rsquared 로 결정계수, params['flipper_length_mm'] 로 기울기, pvalues 로 p값을 꺼낸다
```

</details>

````python
import pandas as pd
import statsmodels.formula.api as smf

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'flipper_length_mm'])
# ols 식은 '종속변수 ~ 설명변수' 로 읽는다. 여기서는 물갈퀴 길이로 체중을 설명한다.
simple_model = smf.ols('body_mass_g ~ flipper_length_mm', data=df).fit()
# R^2 는 설명력, params 는 기울기('1mm 늘 때 몇 g'), pvalues 는 그 기울기가 0 이 아니라는 근거.
simple_r2 = simple_model.rsquared
simple_coef = simple_model.params['flipper_length_mm']
simple_p = simple_model.pvalues['flipper_length_mm']
print(round(simple_r2, 3), round(simple_coef, 3), round(simple_p, 4))
````

````python
# [자가채점]
assert abs(simple_r2 - 0.759) < 0.01
assert abs(simple_coef - 49.686) < 0.01, '물갈퀴 1mm 증가 시 체중이 약 49.686g 늘어난다는 뜻이에요'
assert abs(simple_p - 0.0) < 0.01
print("✅ 문제7 통과!")
````

### (2) 회귀선 그리기 (그래프 — 자가채점 없음)
**요구사항**:

- 직선의 x 는 물갈퀴 최소~최대를 촘촘히 나눈 값, y 는 절편+기울기*x 로 만든다.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'flipper_length_mm'])
simple_model = smf.ols('body_mass_g ~ flipper_length_mm', data=df).fit()
plt.figure(figsize=(8, 6))
ax = sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', color='steelblue', alpha=0.6)
x_line = np.linspace(df['flipper_length_mm'].min(), df['flipper_length_mm'].max(), 100)
# 회귀식 y = 절편 + 기울기 × x 를 그대로 계산해 직선을 만든다.
#  params 에서 꺼내 쓰므로 모델이 바뀌면 그림도 저절로 따라간다.
y_line = simple_model.params['Intercept'] + simple_model.params['flipper_length_mm'] * x_line
sns.lineplot(x=x_line, y=y_line, color='red', linewidth=2, ax=ax)
ax.set_title('물갈퀴 길이와 체중의 단순 선형회귀')
plt.show()
````

### 해설 — 문제 7
- **접근법**: `smf.ols('y ~ x', data=df).fit()` 가 최소제곱으로 회귀식을 찾습니다. R²=0.759 는 물갈퀴 길이가 체중 분산의 약 76%를 설명한다는 뜻이고, 기울기 49.686 은 물갈퀴가 1mm 길수록 체중이 평균 약 49.7g 늘어남을 의미합니다. 계수 p값이 사실상 0 이라 이 관계는 통계적으로 유의합니다.
- **지난 단원 연결**: 단순회귀의 R²(0.759)는 물갈퀴-체중 **상관계수 r(0.871)의 제곱**과 정확히 같습니다(0.871² ≈ 0.759). 상관의 세기를 제곱하면 '설명한 분산 비율'이 됩니다.
- **흔한 실수**: `params`·`pvalues` 는 인덱스가 변수명(`'flipper_length_mm'`)입니다. 절편은 `'Intercept'` 로 꺼냅니다. 회귀선은 `sns.regplot` 으로도 그릴 수 있지만, 여기서는 적합한 계수를 직접 써서 직선을 그렸습니다.

## 8. 변수를 더하면 나아질까 — 다중 선형회귀 (검정)
**배경**: 물갈퀴 길이에 **부리 길이(`bill_length_mm`)** 를 더하면 체중 예측이 더 좋아질까요? **다중 선형회귀**로 두 변수를 함께 넣고, 각 변수의 기여를 계수 p값과 **조정 결정계수(Adj R²)** 로 확인합니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 `body_mass_g`·`flipper_length_mm`·`bill_length_mm` 결측 행을 제거하세요(`dropna`).
- `smf.ols('body_mass_g ~ flipper_length_mm + bill_length_mm', data=df).fit()` 로 적합해 `multi_model` 에 담으세요.
- 조정 결정계수 `multi_adj_r2 = multi_model.rsquared_adj`, 물갈퀴 계수 `coef_flipper` 와 그 p값 `p_flipper`, 부리 계수 `coef_bill` 와 그 p값 `p_bill` 을 각각 꺼내세요.
- 조정 결정계수·계수는 소수 **셋째 자리**, p값은 소수 **넷째 자리**까지 비교합니다.

**예시**
```
round(multi_adj_r2, 3) → 0.759
round(coef_flipper, 3) → 48.145      round(p_flipper, 4) → 0.0
round(coef_bill, 3)    → 6.047       round(p_bill, 4)    → 0.2438
```
<details><summary>힌트</summary>

```text
접근방법:
- ols 식에 독립변수를 + 로 이어 두 개 이상 넣으면 다중회귀가 된다.
- 조정 결정계수는 변수 개수를 벌점으로 반영한 R² 로 rsquared_adj 로 꺼낸다.

세부구현:
1. 세 열의 결측을 제거한다
2. ols 식 'body_mass_g ~ flipper_length_mm + bill_length_mm' 로 적합한다
3. rsquared_adj 로 조정 R², params/pvalues 에서 각 변수의 계수와 p값을 꺼낸다
```

</details>

````python
import pandas as pd
import statsmodels.formula.api as smf

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'flipper_length_mm', 'bill_length_mm'])
# + 로 설명변수를 하나 더 넣는다. 이때는 R^2 가 아니라 조정된 R^2 를 봐야 한다
#  — R^2 는 쓸모없는 변수를 넣어도 무조건 올라가기 때문이다.
multi_model = smf.ols('body_mass_g ~ flipper_length_mm + bill_length_mm', data=df).fit()
multi_adj_r2 = multi_model.rsquared_adj
coef_flipper = multi_model.params['flipper_length_mm']
p_flipper = multi_model.pvalues['flipper_length_mm']
coef_bill = multi_model.params['bill_length_mm']
p_bill = multi_model.pvalues['bill_length_mm']
print(round(multi_adj_r2, 3), round(coef_flipper, 3), round(p_flipper, 4), round(coef_bill, 3), round(p_bill, 4))
````

````python
# [자가채점]
assert abs(multi_adj_r2 - 0.759) < 0.01
assert abs(coef_flipper - 48.145) < 0.01
assert abs(p_flipper - 0.0) < 0.01
assert abs(coef_bill - 6.047) < 0.01
assert abs(p_bill - 0.2438) < 0.01, '부리 길이 계수는 유의하지 않아요(p>0.05)'
print("✅ 문제8 통과!")
````

### 해설 — 문제 8
- **접근법**: 다중회귀는 다른 변수를 고정한 채 각 변수의 순효과를 봅니다. 물갈퀴 계수는 p≈0 으로 여전히 유의하지만, 부리 길이 계수는 p=0.2438 로 유의하지 않습니다 — 즉 물갈퀴가 이미 설명하는 정보에 부리 길이가 더하는 몫이 거의 없습니다. 조정 R² 도 단순회귀(0.758)에서 0.759 로 사실상 그대로여서, 변수를 더한 이득이 미미함을 확인할 수 있습니다.
- **흔한 실수**: 일반 R² 는 변수를 더할수록 무조건 커지므로 모형 비교에는 **조정 R²** 를 봐야 합니다. 유의하지 않은 변수를 넣으면 조정 R² 는 오히려 거의 안 오르거나 내려갑니다.
- **대안·개념**: 계수가 유의하지 않다고 해서 그 변수가 체중과 무관한 것은 아닙니다. 부리 길이도 혼자 넣으면 체중과 유의하게 관련됩니다 — 물갈퀴가 이미 설명한 뒤에 **추가로 설명할 몫**이 적을 뿐입니다. 이렇게 예측변수끼리 겹칠 때 계수가 흔들리는 현상은 참고 교안 `참고_다중공선성.ipynb` 에서 다룹니다.

## 9. 회귀가 타당한가 — 잔차 진단 (그래프)

- **왼쪽(`ax1`)**: x=적합값(`simple_model.fittedvalues`), y=잔차(`simple_model.resid`)의 **산점도**를 그리고, 잔차 0 위치에 빨간 **수평 점선**을 그으세요(`ax1.axhline(0, color='red', linestyle='--')`).
- **오른쪽(`ax2`)**: 잔차의 **Q-Q Plot** 을 `pg.qqplot(simple_model.resid, dist='norm', ax=ax2)` 로 그리세요. 점들이 대각선(빨간 기준선)에 가까울수록 잔차가 정규분포에 가깝다는 뜻입니다.

접근방법:
- 서브플롯 두 개를 나란히 만들어 왼쪽엔 적합값-잔차 산점도, 오른쪽엔 잔차 Q-Q Plot 을 그린다.
- 왼쪽 점들이 기준선 위아래로 패턴 없이 흩어져 있으면 등분산·선형 가정이, 오른쪽 점들이 대각선에 가까우면 정규성 가정이 대체로 만족된 것이다.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pingouin as pg
import statsmodels.formula.api as smf

df = pd.read_csv('../../day09_가설검정_회귀분석/data/penguins.csv').dropna(subset=['body_mass_g', 'flipper_length_mm'])
simple_model = smf.ols('body_mass_g ~ flipper_length_mm', data=df).fit()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
sns.scatterplot(x=simple_model.fittedvalues, y=simple_model.resid, color='steelblue', alpha=0.6, ax=ax1)
# 잔차가 0 선을 중심으로 위아래 고르게 흩어져 있어야 한다.
#  깔때기 모양이면 등분산 위반, 휘어 있으면 관계가 직선이 아니라는 뜻이다.
ax1.axhline(0, color='red', linestyle='--')
ax1.set_title('잔차 vs 적합값 (등분산·선형성 진단)')
ax1.set_xlabel('적합값 (예측 체중)')
ax1.set_ylabel('잔차 (실제 - 예측)')
pg.qqplot(simple_model.resid, dist='norm', ax=ax2)
ax2.set_title('잔차 Q-Q Plot (정규성 진단)')
plt.show()
````

### 해설 — 문제 9
- **접근법**: 왼쪽 잔차 vs 적합값 그림에서 점들이 0 을 중심으로 **특정 패턴(깔때기·곡선) 없이** 고르게 흩어지면 등분산·선형 가정이 대체로 만족됩니다. 펄럭이는 곡선이 보이면 비선형, 오른쪽으로 갈수록 퍼지면 이분산 신호입니다. 오른쪽 잔차 Q-Q Plot 은 점들이 대체로 대각선을 따라가므로 잔차가 정규분포에 크게 벗어나지 않는다는 뜻이며, **Q-Q 가 직선이면 정규성 가정이 충족된 것**으로 읽습니다(양 끝 꼬리에서 살짝 벗어나는 정도는 흔합니다).
- **흔한 실수**: 잔차를 원래 x(물갈퀴 길이)가 아니라 **적합값(예측값)** 에 대해 그립니다. `model.resid` 와 `model.fittedvalues` 는 결측 제거한 데이터 순서와 같은 길이여야 하므로, 그림 전에 같은 `dropna` 를 거친 데이터로 적합하세요. `pg.qqplot` 은 `ax=ax2` 인자를 줘야 그 축에 바로 그려집니다(안 주면 새 figure 가 따로 뜹니다).
- **대안**: 잔차의 독립성은 Durbin-Watson(개념)으로 함께 봅니다(2에 가까우면 자기상관 없음). 회귀 진단은 **LINE 네 가정**(선형성·독립성·정규성·등분산성)을 **세 도구**로 확인합니다 — 잔차 vs 적합값(선형성·등분산성) + Q-Q(정규성) + Durbin-Watson(독립성).

## 10. 인사이트 리포트 (종합 서술)
**배경**: 지금까지의 회귀 결과를 근거로 "체중을 예측할 때 어떤 변수가 중요한가"를 한 문단으로 정리합니다.

**요구사항**:
- 단순회귀(문제 7)와 다중회귀(문제 8)의 결과를 근거로 들며, **어떤 변수가 체중 예측에 기여하는지**를 **3문장 이상**으로 아래 서술 셀에 적으세요.
- 다음을 포함하면 좋아요: 물갈퀴 길이의 설명력(R²·계수 p값), 부리 길이를 더했을 때 조정 R²·계수 p값의 변화, 그리고 상관·회귀는 인과를 증명하지 않는다는 점.

**인사이트 리포트 (모범 서술 — 예시)**

단순회귀에서 물갈퀴 길이(`flipper_length_mm`)만으로 체중 분산의 약 76%(R²=0.759)를 설명했고 계수의 p값도 사실상 0 이라, 물갈퀴 길이는 체중의 강력한 예측 변수다. 다중회귀에 부리 길이(`bill_length_mm`)를 더해도 조정 R² 는 0.758 에서 0.759 로 거의 그대로였고 부리 길이 계수의 p값은 0.2438 로 유의하지 않아, 물갈퀴 길이가 이미 설명하는 정보에 부리 길이가 더하는 몫은 미미하다. 따라서 체중 예측에는 물갈퀴 길이가 핵심 변수이며, 변수를 무작정 늘리기보다 계수 p값과 조정 R² 변화로 기여도를 확인해 선택해야 한다. 다만 이는 상관·회귀에 기반한 설명일 뿐 물갈퀴가 커서 체중이 는다는 인과를 증명하는 것은 아니며, 종·먹이 같은 다른 요인이 함께 작용한다.

### 해설 — 문제 10
- **평가 포인트**: 1) 물갈퀴 길이의 설명력(R²=0.759·계수 p≈0) 2) 부리 길이 추가 시 조정 R² 거의 불변·계수 p=0.2438(유의하지 않음) 3) 그래서 물갈퀴 길이가 핵심 변수라는 결론 4) 상관≠인과 언급 — 이 네 가지를 근거 숫자와 함께 담았는지 봅니다.
- **흔한 실수**: 숫자 없이 "물갈퀴가 중요하다"만 적으면 근거가 약합니다. 앞에서 구한 R²·조정 R²·계수 p값을 인용하도록 지도하세요.
