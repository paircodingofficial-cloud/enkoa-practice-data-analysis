## 2. 평균과 중앙값
**배경**: 데이터의 '가운데'를 나타내는 가장 기본적인 두 대푯값이 평균과 중앙값입니다. 펭귄 몸무게의 평균과 중앙값을 구해 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `body_mass_g` 열의 **결측치를 제거하고**, 평균을 `mass_mean`, 중앙값을 `mass_median` 에 담으세요.
- 값은 각각 소수 첫째 자리까지 봤을 때 채점됩니다(`round(값, 1)`).

**예시**
```
round(mass_mean, 1)    →  4201.8
round(mass_median, 1)  →  4050.0
```
<details><summary>힌트</summary>

```text
접근방법:
- 한 열의 결측을 없앤 뒤 평균과 중앙값을 구한다.

세부구현:
1. 파일을 df 로 불러온다
2. body_mass_g 열에서 결측치를 제거한다(dropna)
3. 남은 값의 평균(mean)과 중앙값(median)을 각각 변수에 담는다
```

````python
import pandas as pd

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
mass = df['body_mass_g'].dropna()
mass_mean = mass.mean()
# 평균과 중앙값을 나란히 본다 — 둘이 크게 벌어지면 분포가 한쪽으로 치우쳤다는 신호다.
mass_median = mass.median()
print("평균:", round(mass_mean, 1), "중앙값:", round(mass_median, 1))
````

````python
# [자가채점]  통계값은 정확일치 대신 허용오차(abs<tol)로 비교합니다.
assert abs(mass_mean - 4201.8) < 0.1
assert abs(mass_median - 4050.0) < 0.1
print("✅ 문제2 통과!")
````

### 해설 — 문제 2
- **접근법**: `dropna()` 로 결측을 없앤 Series 에 `.mean()`·`.median()` 을 씁니다. 평균은 모든 값을 더해 개수로 나눈 값, 중앙값은 크기순으로 세웠을 때 한가운데 값입니다.
- **흔한 실수**: 결측을 제거하지 않아도 pandas 의 `.mean()` 은 자동으로 NaN 을 빼고 계산하지만, 습관적으로 `dropna()` 를 먼저 해 두면 이후 개수 기반 계산에서 실수를 줄일 수 있어요.
- **대안**: 평균이 중앙값보다 크면 분포가 오른쪽으로, 작으면 왼쪽으로 치우친 신호입니다.

## 3. 절사평균 (trim_mean)
**배경**: 평균은 극단값에 휘둘리기 쉽습니다. **절사평균**은 양 끝 일부를 잘라내고 평균을 내 극단값의 영향을 줄인 대푯값입니다. 부리 길이의 절사평균을 구해 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `bill_length_mm` 열의 **결측치를 제거하고**, `stats.trim_mean(값, 0.1)` 로 양 끝 10%씩을 잘라낸 절사평균을 구해 `trimmed_mean` 에 담으세요.
- 값은 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).

**예시**
```
round(trimmed_mean, 2)  →  43.91
```
<details><summary>힌트</summary>

```text
접근방법:
- 결측을 없앤 열을 trim_mean 에 넘기고, 잘라낼 비율로 0.1 을 준다.

세부구현:
1. 파일을 df 로 불러온다
2. bill_length_mm 열에서 결측치를 제거한다(dropna)
3. stats.trim_mean 에 그 값과 비율 0.1 을 주어 trimmed_mean 에 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
bill = df['bill_length_mm'].dropna()
# 0.1 = 양쪽 끝 10%씩 잘라 낸다. 극단값에 덜 흔들리는 평균이 필요할 때 쓴다.
trimmed_mean = stats.trim_mean(bill, 0.1)
print("절사평균:", round(trimmed_mean, 2))
````

````python
# [자가채점]
assert abs(trimmed_mean - 43.91) < 0.01
print("✅ 문제3 통과!")
````

### 해설 — 문제 3
- **접근법**: `stats.trim_mean(x, 0.1)` 은 정렬 후 아래·위 10%씩(합 20%)을 버리고 나머지의 평균을 냅니다. 극단값이 있어도 대푯값이 덜 흔들려요.
- **흔한 실수**: 비율 인자는 **한쪽** 비율입니다. 0.1 을 주면 양 끝을 합쳐 20%가 잘립니다.
- **대안**: 극단값에 가장 강한 대푯값은 중앙값(절사 비율 0.5 에 가까운 극단)입니다.

## 4. 분산과 표준편차

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `flipper_length_mm` 열의 **결측치를 제거하고**, 표본 분산을 `flipper_var`, 표본 표준편차를 `flipper_std` 에 담으세요. (표본 기준이므로 `.var(ddof=1)`·`.std(ddof=1)` 을 쓰세요.)
- 값은 각각 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).

**예시**
```
round(flipper_var, 2)  →  197.73

```
<details><summary>힌트</summary>

```text
접근방법:
- 결측을 없앤 열에 분산·표준편차를 구하되, 표본 기준(ddof=1)으로 계산한다.

세부구현:
1. 파일을 df 로 불러온다
2. flipper_length_mm 열에서 결측치를 제거한다(dropna)
3. var 와 std 를 각각 ddof 를 1 로 주어 구하고 변수에 담는다
```

````python
import pandas as pd

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
flipper = df['flipper_length_mm'].dropna()
# ddof=1 은 표본 기준(n-1 로 나눔). 표본으로 모집단을 추정할 때의 기본값이다.
#  pandas 는 기본이 1 이지만 numpy 는 0 이라, 두 결과가 달라 보이는 흔한 원인이다.
flipper_var = flipper.var(ddof=1)
flipper_std = flipper.std(ddof=1)
print("분산:", round(flipper_var, 2), "표준편차:", round(flipper_std, 2))
````

````python
# [자가채점]
assert abs(flipper_var - 197.73) < 0.01
assert abs(flipper_std - 14.06) < 0.01
print("✅ 문제4 통과!")
````

### 해설 — 문제 4
- **접근법**: 분산은 각 값이 평균에서 벗어난 정도를 제곱해 평균한 값, 표준편차는 그 제곱근입니다. 표준편차는 원래 단위(mm)로 돌아와 해석하기 쉬워요.
- **흔한 실수**: `ddof` 를 빼면 pandas 는 이미 기본이 `ddof=1`(표본)이지만, numpy 의 `np.var`·`np.std` 는 기본이 `ddof=0`(모집단)이라 값이 달라집니다. 도구마다 기본값을 확인하세요.
- **대안**: 표준편차를 평균으로 나누면 다음 문제의 변동계수(CV)가 됩니다.

## 5. IQR과 이상치
**배경**: IQR(사분위 범위)은 3사분위수(Q3)에서 1사분위수(Q1)를 뺀 값으로, 가운데 50%가 퍼진 폭입니다. **1.5×IQR 규칙**으로 몸무게에 이상치가 몇 개나 있는지 세어 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `body_mass_g` 열의 **결측치를 제거하고**, `stats.iqr(값)` 으로 IQR 을 구해 `mass_iqr` 에 담으세요.
- Q1·Q3 는 `.quantile(0.25)`·`.quantile(0.75)` 로 구하고, 정상 범위의 아래 경계 **(Q1 − 1.5×IQR)** 를 `lower_bound`, 위 경계 **(Q3 + 1.5×IQR)** 를 `upper_bound` 에 담으세요.
- `lower_bound` **미만**이거나 `upper_bound` **초과**인 값을 이상치로 보고, 그 **개수**를 정수로 `outlier_count` 에 담으세요.
- `mass_iqr`·`lower_bound`·`upper_bound` 는 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).

**예시**
```
round(mass_iqr, 2)     →  1200.0
round(lower_bound, 2)  →  1750.0
round(upper_bound, 2)  →  6550.0
outlier_count          →  0
```
> 이 규칙으로 걸리는 값이 **0개**여도 코드가 틀린 게 아닙니다 — 그 자체가 결과예요. 다른 규칙(Z-score)으로 세면 결과가 달라지는데, 그건 문제 15 에서 확인합니다.
<details><summary>힌트</summary>

```text
접근방법:
- IQR 로 정상 범위(아래 경계~위 경계)를 만들고, 그 밖의 값을 센다.

세부구현:
1. 파일을 df 로 불러오고 body_mass_g 결측을 제거한다(dropna)
2. stats.iqr 로 IQR 을, quantile 로 Q1·Q3 을 구한다
3. 아래 경계·위 경계를 각각 변수에 담고, 그 밖에 있는 값의 개수를 세어 정수로 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
mass = df['body_mass_g'].dropna()
# IQR = Q3 - Q1. 가운데 50%가 퍼진 폭이라 극단값에 흔들리지 않는다.
mass_iqr = stats.iqr(mass)
q1 = mass.quantile(0.25)
q3 = mass.quantile(0.75)
lower_bound = q1 - 1.5 * mass_iqr
upper_bound = q3 + 1.5 * mass_iqr
outlier_count = int(((mass < lower_bound) | (mass > upper_bound)).sum())
print("IQR:", round(mass_iqr, 2), "정상범위:", round(lower_bound, 2), "~", round(upper_bound, 2))
print("IQR 규칙 이상치:", outlier_count)
````

````python
# [자가채점]
assert abs(mass_iqr - 1200.0) < 0.01
assert abs(lower_bound - 1750.0) < 0.01
assert abs(upper_bound - 6550.0) < 0.01
assert outlier_count == 0
print("✅ 문제5 통과!")
````

### 해설 — 문제 5
- **접근법**: IQR 은 가운데 50%의 폭이라 극단값에 강합니다. 정상 범위를 `[Q1 − 1.5×IQR, Q3 + 1.5×IQR]` = `[1750, 6550]` 로 잡고 그 밖을 이상치로 봅니다(상자그림의 수염 규칙과 같아요).
- **두 규칙이 왜 다른가**: 펭귄 몸무게는 2700~6300g 범위라 `[1750, 6550]` 을 벗어나는 값이 **하나도 없습니다(0개)**. 반면 Z-score 규칙(`|Z| > 2`)은 평균·표준편차 기준이라 분포의 양 끝 **9마리**를 잡아냅니다. **이상치는 데이터가 정해 주는 것이 아니라 규칙이 정의하는 것**이라는 게 이 문제의 교훈이에요. 어느 쪽이 옳다기보다, 목적에 맞는 규칙을 **밝히고** 쓰는 것이 중요합니다.
- **0 은 실패가 아니다**: IQR 규칙 결과가 0 이라고 코드가 틀린 게 아닙니다. **이 데이터에 IQR 기준 극단값이 없다**는 사실 자체가 결과예요(penguins 는 네 수치 열 모두 IQR 이상치가 0개입니다).
- **흔한 실수**: 두 조건을 `|`(또는) 가 아니라 `&`(그리고) 로 이으면 "아래 경계보다 작으면서 동시에 위 경계보다 크다"는 불가능한 조건이 되어 **언제나 0** 이 나옵니다. 경계식을 아무리 틀리게 써도 0 이 나오므로 스스로 알아채기 어렵죠 — 그래서 이 문제는 경계값과 Z-score 개수까지 함께 채점합니다.
- **대안**: `.describe()` 의 25%·75% 값으로 직접 Q1·Q3 을 확인해도 같은 결과가 나옵니다. `|Z| > 3` 으로 기준을 높이면 이 데이터에서는 0개가 됩니다.

## 6. 변동계수 (CV)
**배경**: 단위가 다른 두 변수의 산포를 견주려면 표준편차만으로는 부족합니다. **변동계수(CV) = 표준편차 ÷ 평균 × 100 (%)** 은 평균 대비 상대적 흩어짐이라 단위와 무관하게 비교됩니다. 부리 길이와 몸무게 중 어느 쪽이 상대적으로 더 퍼져 있는지 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `bill_length_mm` 와 `body_mass_g` 를 **각각 결측 제거**한 뒤, 변동계수를 **백분율(%)** 로 — 즉 `표준편차 ÷ 평균 × 100` (표준편차는 `.std()`, 표본 기준) — 구해 각각 `cv_bill`, `cv_mass` 에 담으세요.
- 두 CV 중 **더 큰 쪽의 열 이름**('bill_length_mm' 또는 'body_mass_g')을 문자열로 `larger_cv` 에 담으세요.
- `cv_bill`·`cv_mass` 는 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).

**예시**
```
round(cv_bill, 2)  →  12.43
round(cv_mass, 2)  →  19.09
larger_cv          →  'body_mass_g'
```
<details><summary>힌트</summary>

```text
접근방법:
- 각 열의 표준편차를 평균으로 나눠 CV 를 구하고, 둘을 비교한다.

세부구현:
1. 파일을 df 로 불러오고 두 열을 각각 결측 제거한다(dropna)
2. 각 열의 표준편차(std)를 평균(mean)으로 나눈 뒤 100 을 곱해 CV(%)를 구한다
3. 두 CV 를 비교해 더 큰 쪽의 열 이름 문자열을 larger_cv 에 담는다
```

</details>

````python
import pandas as pd

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
bill = df['bill_length_mm'].dropna()
mass = df['body_mass_g'].dropna()
# 변동계수 = 표준편차 ÷ 평균. 단위가 다른 두 변수(mm 와 g)의 흩어진 정도를 견주려면 이렇게 비율로 만든다.
cv_bill = bill.std() / bill.mean() * 100
cv_mass = mass.std() / mass.mean() * 100
larger_cv = 'bill_length_mm' if cv_bill > cv_mass else 'body_mass_g'
print("bill CV:", round(cv_bill, 2), "mass CV:", round(cv_mass, 2), "더 큰 쪽:", larger_cv)
````

````python
# [자가채점]
assert abs(cv_bill - 12.43) < 0.01
assert abs(cv_mass - 19.09) < 0.01
assert larger_cv == 'body_mass_g'
print("✅ 문제6 통과!")
````

### 해설 — 문제 6
- **접근법**: CV 는 표준편차를 평균으로 나눈 **상대 산포**입니다. mm 와 g 처럼 단위가 다른 두 변수도 같은 잣대로 흩어짐을 비교할 수 있어요.
- **흔한 실수**: 표준편차(절대 산포)만 보면 값이 큰 몸무게가 무조건 더 퍼져 보입니다. 평균으로 나눠야 공정한 비교가 됩니다.
- **대안**: 100 을 곱하지 않고 비율(0.1243·0.1909)로 두기도 합니다. 교안과 LV2·LV3 는 이 문제와 같은 **백분율(%)** 표기를 쓰니 자료 안에서는 % 로 통일해 두세요. 비교 결과(어느 쪽이 큰지)는 어느 표기든 같습니다.

## 7. 왜도와 첨도
**배경**: 분포의 '모양'을 숫자로 요약한 값이 왜도(치우침)와 첨도(뾰족함)입니다. 왜도가 0보다 크면 오른쪽 꼬리가 길고, 작으면 왼쪽 꼬리가 깁니다. 부리 깊이의 분포 모양을 재 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `bill_depth_mm` 열의 **결측치를 제거하고**, `stats.skew(값)` 으로 왜도를 `depth_skew`, `stats.kurtosis(값)` 으로 첨도를 `depth_kurt` 에 담으세요.
- 값은 각각 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).

**예시**
```
round(depth_skew, 3)  →  -0.143
round(depth_kurt, 3)  →  -0.911
```
<details><summary>힌트</summary>

```text
접근방법:
- 결측을 없앤 열을 skew·kurtosis 에 각각 넘긴다.

세부구현:
1. 파일을 df 로 불러온다
2. bill_depth_mm 열에서 결측치를 제거한다(dropna)
3. stats.skew 와 stats.kurtosis 로 왜도·첨도를 구해 변수에 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
depth = df['bill_depth_mm'].dropna()
# 왜도는 좌우 치우침(0 이면 대칭), 첨도는 꼬리의 두께(scipy 는 정규분포를 0 으로 맞춘 초과첨도다).
depth_skew = stats.skew(depth)
depth_kurt = stats.kurtosis(depth)
print("왜도:", round(depth_skew, 3), "첨도:", round(depth_kurt, 3))
````

````python
# [자가채점]
assert abs(depth_skew - (-0.143)) < 0.01
assert abs(depth_kurt - (-0.911)) < 0.01
print("✅ 문제7 통과!")
````

### 해설 — 문제 7
- **접근법**: `stats.skew` 는 좌우 치우침을, `stats.kurtosis` 는 뾰족함(꼬리 두께)을 잰다. 부리 깊이는 왜도가 0 에 가깝고(거의 대칭) 첨도가 음수라 정규분포보다 완만한 편입니다.
- **흔한 실수**: `stats.kurtosis` 는 기본이 **초과첨도**(정규분포=0 기준)입니다. 정규분포=3 기준의 값을 원하면 `fisher=False` 를 줘야 해요(여기선 기본값 사용).
- **대안**: 히스토그램·KDE 를 함께 그려 보면 이 숫자들이 뜻하는 모양을 눈으로 확인할 수 있습니다.

## 9. 피어슨 상관계수
**배경**: 두 수치 변수의 **직선 관계 강도**를 −1~+1 사이 한 숫자로 요약한 값이 피어슨 상관계수입니다. 날개 길이와 몸무게가 얼마나 함께 커지는지 재 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `flipper_length_mm` 와 `body_mass_g` 두 열을 함께 **결측 제거**한 뒤, `stats.pearsonr` 로 상관계수를 구해 `pearson_r` 에 담으세요. (`stats.pearsonr` 는 **두 값**을 돌려주니 **첫 번째(상관계수)** 만 쓰세요.)
- 상관계수의 절댓값이 **0.7 이상이면 '강함'**, **0.3 이상 0.7 미만이면 '중간'**, **0.3 미만이면 '약함'** 으로 판단해 문자열을 `strength` 에 담으세요.
- `pearson_r` 는 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).

**예시**
```
round(pearson_r, 3)  →  0.871
strength             →  '강함'
```
<details><summary>힌트</summary>

```text
접근방법:
- 두 열을 함께 결측 제거한 뒤 pearsonr 로 상관계수를 구하고, 절댓값 크기로 강도를 나눈다.

세부구현:
1. 파일을 df 로 불러오고 두 열만 골라 함께 결측 제거한다(dropna)
2. pearsonr 에 두 열을 넘겨 첫 번째 반환값(상관계수)을 pearson_r 에 담는다
3. 상관계수의 절댓값을 기준값(0.7·0.3)과 견주어 강함·중간·약함 문자열을 strength 에 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
pair = df[['flipper_length_mm', 'body_mass_g']].dropna()
# [0] 은 상관계수(뒤의 [1] 은 p 값이다). 피어슨은 '직선' 관계의 세기를 잰다.
pearson_r = stats.pearsonr(pair['flipper_length_mm'], pair['body_mass_g'])[0]
if abs(pearson_r) >= 0.7:
    strength = '강함'
elif abs(pearson_r) >= 0.3:
    strength = '중간'
else:
    strength = '약함'
print("피어슨 r:", round(pearson_r, 3), "강도:", strength)
````

````python
# [자가채점]
assert abs(pearson_r - 0.871) < 0.01
assert strength == '강함'
print("✅ 문제9 통과!")
````

### 해설 — 문제 9
- **접근법**: 피어슨 상관은 두 변수의 **직선 관계**를 −1~+1 로 요약합니다. 0.871 은 +1 에 가까워, 날개가 길수록 몸무게도 함께 무거워지는 강한 양의 관계입니다.
- **흔한 실수**: `pearsonr` 는 값 두 개를 돌려줍니다. `r, p = stats.pearsonr(...)` 처럼 첫 번째만 상관계수로 받으세요. 결측이 섞이면 에러가 나니 두 열을 함께 `dropna` 해야 합니다.
- **대안**: 여러 열의 상관을 한 번에 보려면 `df.corr(numeric_only=True)` 로 상관행렬을 만듭니다(문제 11).

## 10. 스피어만 상관과 '상관 ≠ 인과'
**배경**: 피어슨이 직선 관계라면 **스피어만 상관**은 값을 <strong>순위(등수)</strong>로 바꿔 계산해 '한쪽이 커질 때 다른 쪽도 커지는' 단조 관계를 봅니다. 이상치나 곡선 관계에 덜 민감합니다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `flipper_length_mm` 와 `body_mass_g` 두 열을 함께 **결측 제거**한 뒤, `stats.spearmanr` 로 스피어만 상관계수를 구해 `spearman_rs` 에 담으세요. (역시 **두 값**을 돌려주니 **첫 번째(상관계수)** 만 쓰세요.)
- 값은 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).
- 그리고 **아래 서술 셀**에, 날개 길이와 몸무게의 상관이 강하다고 해서 '날개가 길어서 몸무게가 늘었다'고 단정할 수 있을지, **상관과 인과의 차이**를 1~2문장으로 적으세요.

**예시**
```
round(spearman_rs, 3)  →  0.84
```
<details><summary>힌트</summary>

```text
접근방법:
- 두 열을 함께 결측 제거한 뒤 spearmanr 로 순위 기반 상관을 구한다.

세부구현:
1. 파일을 df 로 불러오고 두 열만 골라 함께 결측 제거한다(dropna)
2. spearmanr 에 두 열을 넘겨 첫 번째 반환값을 spearman_rs 에 담는다
3. 서술 셀에 상관과 인과의 차이를 한두 문장으로 적는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
pair = df[['flipper_length_mm', 'body_mass_g']].dropna()
# 스피어만은 값 대신 순위로 계산한다 — 곡선 관계나 극단값이 있어도 잘 버틴다.
spearman_rs = stats.spearmanr(pair['flipper_length_mm'], pair['body_mass_g'])[0]
print("스피어만 rs:", round(spearman_rs, 3))
````

````python
# [자가채점]
assert abs(spearman_rs - 0.84) < 0.01
print("✅ 문제10 통과!")
````

**모범 서술 (예시)**

날개 길이와 몸무게는 스피어만 상관도 약 0.84 로 매우 강하지만, 이것만으로 '날개가 길어서 몸무게가 늘었다'고 **인과관계**를 단정할 수는 없다. 두 값은 모두 '몸집이 큰 개체일수록 함께 커지는' 공통 요인(종·나이·성별 등)의 영향을 받았을 수 있기 때문이다. **상관은 '함께 움직인다'는 사실일 뿐, 무엇이 원인인지는 말해 주지 않는다.**

## 12. 정규분포로 확률 구하기
**배경**: 데이터가 대략 정규분포를 따른다고 가정하면, 평균과 표준편차만으로 '어떤 값 이하일 확률'을 계산할 수 있습니다. 몸무게가 정규분포를 따른다고 보고, 4000g 이하일 확률을 구해 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `body_mass_g` 열의 **결측치를 제거하고**, 평균과 표준편차(`.std()`, 표본 기준)를 구하세요.
- 이 평균·표준편차로 정규분포를 가정하고, `stats.norm.cdf(4000, 평균, 표준편차)` 로 몸무게가 **4000g 이하일 확률**을 구해 `prob_under_4000` 에 담으세요.
- 값은 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).

**예시**
```
round(prob_under_4000, 3)  →  0.401
```
<details><summary>힌트</summary>

```text
접근방법:
- 몸무게의 평균·표준편차를 구해 정규분포를 세우고, cdf 로 특정 값 이하의 누적 확률을 읽는다.

세부구현:
1. 파일을 df 로 불러오고 body_mass_g 결측을 제거한다(dropna)
2. 평균(mean)과 표준편차(std)를 구한다
3. norm.cdf 에 4000 과 평균·표준편차를 넘겨 4000 이하일 확률을 prob_under_4000 에 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
mass = df['body_mass_g'].dropna()
mu = mass.mean()
sigma = mass.std()
# cdf 는 '그 값 이하일 확률' — 왼쪽 꼬리부터 누적한 넓이다.
prob_under_4000 = stats.norm.cdf(4000, mu, sigma)
print("4000g 이하일 확률:", round(prob_under_4000, 3))
````

````python
# [자가채점]
assert abs(prob_under_4000 - 0.401) < 0.01
print("✅ 문제12 통과!")
````

### 해설 — 문제 12
- **접근법**: `norm.cdf(x, 평균, 표준편차)` 는 정규분포에서 **x 이하의 누적 확률**을 돌려줍니다. 평균(약 4202g)보다 4000g 이 조금 작으니 확률이 0.5 보다 약간 작은 0.401 이 나옵니다.
- **흔한 실수**: `norm.cdf` 에 평균·표준편차를 넘기지 않으면 표준정규분포(평균 0, 표준편차 1)로 계산해 엉뚱한 값이 나옵니다. 반드시 데이터의 평균·표준편차를 함께 넘기세요.
- **대안**: '4000g 초과일 확률'은 `1 - prob_under_4000`, '두 값 사이 확률'은 두 cdf 의 차로 구합니다.

## 13. 이항분포로 확률 계산
**배경**: 어느 관찰소의 기록에 따르면, 무작위로 마주치는 펭귄이 Adelie 종일 확률은 대략 `p=0.44` 입니다(전체 344마리 중 Adelie 152마리 비율과 비슷한 값입니다). 이 확률로 펭귄 10마리를 독립적으로 관찰한다고 할 때, 그중 Adelie 가 몇 마리 나올지는 **이항분포**로 계산할 수 있습니다.

**요구사항**:
- 데이터 로드 없이 아래 값만으로 계산합니다: 시행 횟수 `n=10`, 성공 확률 `p=0.44`.
- `stats.binom.pmf(4, 10, 0.44)` 로 **정확히 4마리**가 Adelie 일 확률을 구해 `p_exactly_4` 에 담으세요.
- `stats.binom.cdf(4, 10, 0.44)` 로 **4마리 이하**가 Adelie 일 확률을 구해 `p_at_most_4` 에 담으세요.
- 두 값 모두 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).

**예시**
```
round(p_exactly_4, 3)  →  0.243
round(p_at_most_4, 3)  →  0.53
```
<details><summary>힌트</summary>

```text
접근방법:
- 이항분포는 성공확률 p 인 시행을 n 번 반복할 때 성공 횟수의 분포다. pmf 는 '정확히', cdf 는 '이하'.

세부구현:
1. binom.pmf 에 성공 횟수 4, 시행 횟수 10, 성공확률 0.44 를 순서대로 넘겨 p_exactly_4 에 담는다
2. binom.cdf 에 같은 순서(4, 10, 0.44)로 넘겨 p_at_most_4 에 담는다
```

</details>

````python
from scipy import stats

# pmf 는 '정확히 그 값', cdf 는 '그 값 이하' — 이항분포처럼 셀 수 있는 값에서만 pmf 를 쓴다.
p_exactly_4 = stats.binom.pmf(4, 10, 0.44)
p_at_most_4 = stats.binom.cdf(4, 10, 0.44)
print("정확히 4마리:", round(p_exactly_4, 3), "4마리 이하:", round(p_at_most_4, 3))
````

````python
# [자가채점]
assert abs(p_exactly_4 - 0.243) < 0.01
assert abs(p_at_most_4 - 0.53) < 0.01
print("✅ 문제13 통과!")
````

### 해설 — 문제 13
- **접근법**: 이항분포는 '성공/실패' 두 결과만 있는 시행을 독립적으로 n 번 반복할 때 성공 횟수의 분포입니다. `pmf(k, n, p)` 는 **정확히 k 번** 성공할 확률, `cdf(k, n, p)` 는 **k 번 이하** 성공할 누적 확률입니다.
- **흔한 실수**: 인자 순서를 헷갈리기 쉽습니다. `binom.pmf(성공횟수, 시행횟수, 성공확률)` 순서를 지키세요 — `pmf(10, 4, 0.44)` 처럼 뒤바꾸면 전혀 다른 값이 나옵니다.
- **대안**: '4마리 초과'일 확률은 `1 - p_at_most_4`, '정확히 5마리 이상 8마리 이하'처럼 구간 확률은 `binom.cdf(8,...) - binom.cdf(4,...)` 로 구합니다.

## 14. 포아송분포로 확률 계산
**배경**: 어느 펭귄 관측소에는 한 시간에 평균 `λ=3` 마리의 펭귄이 도착합니다(관측 기록으로 추정한 평균 도착 수). 한 시간 동안 도착하는 펭귄 수처럼, **평균 발생 횟수만 알면** 일정 시간 동안 몇 번 일어날지는 **포아송분포**로 계산할 수 있습니다.

**요구사항**:
- 데이터 로드 없이 아래 값만으로 계산합니다: 평균 발생 횟수 `λ=3`.
- `stats.poisson.pmf(5, 3)` 으로 한 시간 동안 **정확히 5마리**가 도착할 확률을 구해 `p_exactly_5` 에 담으세요.
- `stats.poisson.cdf(2, 3)` 으로 한 시간 동안 **2마리 이하**가 도착할 확률을 구해 `p_at_most_2` 에 담으세요.
- 두 값 모두 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).

**예시**
```
round(p_exactly_5, 3)  →  0.101
round(p_at_most_2, 3)  →  0.423
```
<details><summary>힌트</summary>

```text
접근방법:
- 포아송분포는 평균 발생 횟수(λ)만으로 일정 구간 동안의 발생 횟수를 다룬다. pmf 는 '정확히', cdf 는 '이하'.

세부구현:
1. poisson.pmf 에 발생 횟수 5, 평균 발생 횟수 3 을 순서대로 넘겨 p_exactly_5 에 담는다
2. poisson.cdf 에 같은 순서(2, 3)로 넘겨 p_at_most_2 에 담는다
```

````python
from scipy import stats

# 포아송은 '단위 시간·공간당 평균 3번 일어나는 일'을 다룬다. 인자 3 이 그 평균이다.
p_exactly_5 = stats.poisson.pmf(5, 3)
p_at_most_2 = stats.poisson.cdf(2, 3)
print("정확히 5마리:", round(p_exactly_5, 3), "2마리 이하:", round(p_at_most_2, 3))
````

````python
# [자가채점]
assert abs(p_exactly_5 - 0.101) < 0.01
assert abs(p_at_most_2 - 0.423) < 0.01
print("✅ 문제14 통과!")
````

### 해설 — 문제 14

- **대안**: 시행 횟수가 매우 크고 성공확률이 매우 작은 이항분포는 포아송분포로 근사됩니다 — '드물게 일어나는 사건의 개수'를 다룰 때 포아송을 씁니다.

## 15. Z-score와 경험 규칙
**배경**: 특정 값이 평균에서 표준편차 몇 배만큼 떨어져 있는지를 **Z-score** 로 나타냅니다. 정규분포에서는 평균 ± 1표준편차 구간에 전체의 약 68%가 몰려 있다는 **경험 규칙(68-95-99.7)** 도 있습니다. 몸무게 5000g인 펭귄이 얼마나 특이한 값인지 Z-score 로 확인해 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `body_mass_g` 열의 **결측치를 제거하고**, 평균과 표준편차(`.std()`, 표본 기준)를 구하세요.
- 몸무게 5000g 의 Z-score, 즉 `(5000 − 평균) ÷ 표준편차` 를 계산해 `z_5000` 에 담으세요. 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).
- `stats.norm.cdf(1) − stats.norm.cdf(-1)` 로, 표준정규분포에서 **평균 ±1표준편차 구간에 들 확률**(경험 규칙의 68%에 해당)을 구해 `prob_within_1sd` 에 담으세요. 소수 셋째 자리까지 봤을 때 채점됩니다(`round(값, 3)`).
- Z-score 는 **이상치를 가려내는 두 번째 규칙**이기도 합니다. 몸무게 전체의 Z-score 를 구해 **절댓값이 2보다 큰** 값의 개수를 정수로 `z_outlier_count` 에 담으세요.

**예시**
```
round(z_5000, 2)          →  1.0
round(prob_within_1sd, 3) →  0.683
z_outlier_count           →  9
```
<details><summary>힌트</summary>

```text
접근방법:
- Z-score 는 값이 평균에서 표준편차 몇 배 떨어져 있는지를 나타낸다. 표준정규분포의 cdf 차이로 구간 확률을 구한다.

세부구현:
1. 파일을 df 로 불러오고 body_mass_g 결측을 제거한다(dropna)
2. 평균(mean)과 표준편차(std)를 구한다
3. (5000 − 평균) 을 표준편차로 나눠 z_5000 에 담는다
4. norm.cdf(1) 에서 norm.cdf(-1) 을 빼 prob_within_1sd 에 담는다
5. 몸무게 전체를 표준화(값에서 평균을 빼고 표준편차로 나눔)한 뒤, 절댓값이 2를 넘는 것의 개수를 센다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
mass = df['body_mass_g'].dropna()
mu = mass.mean()
sigma = mass.std()
# Z-score = (값 - 평균) ÷ 표준편차. 단위를 지우고 '평균에서 몇 표준편차'로 바꾼다.
z_5000 = (5000 - mu) / sigma
prob_within_1sd = stats.norm.cdf(1) - stats.norm.cdf(-1)

z = (mass - mass.mean()) / mass.std(ddof=1)
z_outlier_count = int((z.abs() > 2).sum())
print("Z-score:", round(z_5000, 2), "평균±1표준편차 확률:", round(prob_within_1sd, 3))
print("|Z|>2 인 이상치 개수:", z_outlier_count, " (문제 5 의 IQR 규칙은 0개였습니다)")
````

````python
# [자가채점]
assert abs(z_5000 - 1.0) < 0.01
assert abs(prob_within_1sd - 0.683) < 0.01
assert z_outlier_count == 9
print("✅ 문제15 통과!")
````

### 해설 — 문제 15
- **접근법**: Z-score `(x − 평균) ÷ 표준편차` 는 값을 '평균에서 표준편차 몇 배 떨어졌는지'로 바꿔 줍니다. 5000g 의 Z-score 가 약 1.0 이라는 것은 평균보다 표준편차 1개만큼 무겁다는 뜻입니다. 경험 규칙에 따르면 평균 ±1표준편차 구간(Z가 -1~1)에 전체의 약 68%가 들어 있어요.
- **흔한 실수**: `norm.cdf(1)` 은 '표준정규분포에서 1 이하일 확률'이라 그 자체로 68% 가 아닙니다. ±1 구간의 확률을 구하려면 `cdf(1) − cdf(-1)` 처럼 **양쪽 끝의 차**를 계산해야 합니다.
- **대안**: 평균 ±2표준편차는 약 95%, ±3표준편차는 약 99.7% — 이것이 68-95-99.7 규칙입니다.

## 16. 정규분포 커트라인 구하기 (norm.ppf)
**배경**: 문제 12의 `norm.cdf` 가 '값 → 확률'(이 값 이하일 확률)을 구했다면, `norm.ppf` 는 그 반대로 '확률 → 값'을 구합니다. 즉 **누적확률이 어떤 값이 되는 경계 지점**을 되돌려 줍니다. 몸무게가 정규분포를 따른다고 가정하고, **상위 10%에 드는 커트라인 몸무게**(이 값보다 무거우면 상위 10%)를 구해 봅시다.

**요구사항**:
- `data/penguins.csv` 를 읽어 변수 `df` 에 담으세요.
- `body_mass_g` 열의 **결측치를 제거하고**, 평균과 표준편차(`.std()`, 표본 기준)를 구하세요.
- 상위 10% 커트라인은 **하위 90%의 경계값**과 같으므로, `stats.norm.ppf(0.90, 평균, 표준편차)` 로 구해 `cutoff_top10` 에 담으세요.
- 값은 소수 둘째 자리까지 봤을 때 채점됩니다(`round(값, 2)`).

**예시**
```
round(cutoff_top10, 2)  →  5229.5
```

```text
접근방법:
- 몸무게의 평균·표준편차로 정규분포를 세우고, ppf 로 누적확률 0.90 이 되는 경계값을 되돌려 받는다.
- 상위 10% 의 경계는 하위 90% 의 경계와 같으므로 ppf 에 0.90 을 넣는다.

세부구현:
1. 파일을 df 로 불러오고 body_mass_g 결측을 제거한다(dropna)
2. 평균(mean)과 표준편차(std)를 구한다
3. norm.ppf 에 0.90 과 평균·표준편차를 넘겨 cutoff_top10 에 담는다
```

</details>

````python
import pandas as pd
from scipy import stats

df = pd.read_csv('../../day08_기술통계_추론통계/data/penguins.csv')
mass = df['body_mass_g'].dropna()
mu = mass.mean()
sigma = mass.std()
# ppf 는 cdf 의 반대다 — 확률을 주면 그에 해당하는 값을 돌려준다(상위 10% = 누적 0.90 지점).
cutoff_top10 = stats.norm.ppf(0.90, mu, sigma)
print("상위 10% 커트라인:", round(cutoff_top10, 2))
````

````python
# [자가채점]
assert abs(cutoff_top10 - 5229.5) < 0.1
print("✅ 문제16 통과!")
````

### 해설 — 문제 16
- **접근법**: `norm.ppf(q, 평균, 표준편차)` 는 정규분포에서 **누적확률이 q 가 되는 값**을 돌려줍니다. 상위 10% 는 곧 하위 90% 의 경계이므로 `q=0.90` 을 넣으면 약 5229.5g 이 나옵니다 — 이보다 무거우면 상위 10% 예요.
- **흔한 실수**: '상위 10%' 라고 `0.10` 을 넣으면 하위 10% 커트라인(가벼운 쪽)이 나옵니다. 상위 p% 의 경계는 `ppf(1 - p)`, 즉 상위 10% 는 `ppf(0.90)` 입니다.
- **대안**: `norm.cdf` 와 `norm.ppf` 는 서로 역함수입니다 — `cdf(cutoff_top10, 평균, 표준편차)` 를 다시 구하면 약 0.90 이 되어 두 함수가 반대 방향임을 확인할 수 있습니다.
